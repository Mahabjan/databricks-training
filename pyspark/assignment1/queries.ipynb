```python

from pyspark.sql.functions import *

# 1. Create bonus column as 10% of salary
display(df.withColumn("bonus", df.salary * 0.10))
```

```python
# 2. Create annual_salary column
display(df.withColumn("annual_salary", df.salary * 12))
```

```python
# 3. Create tax column as 5% of salary
display(df.withColumn("tax", df.salary * 0.05))
```

```python
# 4. Create updated_salary by adding 5000
display(df.withColumn("updated_salary", df.salary + 5000))
```

```python
# 5. Create salary_category column
display(
    df.withColumn(
        "salary_category",
        when(df.salary >= 80000, "High")
        .when(df.salary >= 50000, "Medium")
        .otherwise("Low")
    )
)
```

```python

# 6. Create age_group column
display(
    df.withColumn(
        "age_group",
        when(df.age < 25, "Young")
        .otherwise("Adult")
    )
)
```

```python
# 7. Create location column by combining city and department
display(
    df.withColumn(
        "location",
        concat_ws("-", df.city, df.department)
    )
)
```

```python
# 8. Create increment_salary column with 15% hike
display(df.withColumn("increment_salary", df.salary * 1.15))
```

```python
# 9. Create experience_status column based on joining year
display(
    df.withColumn(
        "experience_status",
        when(year("joining_date") < 2021, "Experienced")
        .otherwise("New")
    )
)
```

```python
# 10. Create name_length column using employee name
display(df.withColumn("name_length", length(df.emp_name)))
```

```python
# 11. Create is_high_salary column using condition
display(
    df.withColumn(
        "is_high_salary",
        when(df.salary > 80000, True).otherwise(False)
    )
)
```

```python
# 12. Create joining_year column from joining_date
display(df.withColumn("joining_year", year("joining_date")))
```

```python
# 13. Create salary_after_tax column
display(
    df.withColumn(
        "salary_after_tax",
        df.salary - (df.salary * 0.05)
    )
)
```

```python
# 14. Create department_code column
display(
    df.withColumn(
        "department_code",
        substring(df.department, 1, 3)
    )
)
```

```python
# 15. Create double_salary column
display(df.withColumn("double_salary", df.salary * 2))
```
